#### Now we will merge columns overviw,genres,cast and crew into Tags. By converting themm into a same data type.



```python
movies.iloc[0].genres
```

```python
import ast # this is use to convert strings into list
```

```python
def transform(obj):
    lst = []
    for i in ast.literal_eval(obj):
        lst.append(i['name'])
    return lst
```

```python
movies['genres'] = movies['genres'].apply(transform)
```

```python
movies.head()
```

#### Now we dont need all the cast members so we will take only first three cast members. for that we will create a function

```python
def transform_1(obj):
    lst = []
    count = 0
    for i in ast.literal_eval(obj):
        if count != 3:
            lst.append(i['name'])
            count += 1
        else:
            break
    return lst
```

```python
movies['cast'] = movies['cast'].apply(transform_1)
```

```python
## we will create a function to fetch director from crew.

def fetch_director(obj):
    lst = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            lst.append(i['name'])
            break
    return lst
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies.head()
```

```python
## here overview column is in string format so we will convert it into list.append

movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies.head()
```

```python
## here there are lots of spaces between names so we will have to remove that.

movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","")for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies.head()
```

```python
## Now we will concatenate overview, genres, cast and crew into Tags

movies['Tags'] = movies['overview'] + movies['genres'] + movies['cast'] + movies['crew']
```

```python
movies.head()
```

```python
new_dataframe = movies[['id','title','Tags']]
```

```python
new_dataframe
```

```python
new_dataframe['Tags'] = new_dataframe['Tags'].apply(lambda x:" ".join(x))
```

```python
new_dataframe['Tags'] = new_dataframe['Tags'].apply(lambda x:x.lower())
```

```python
new_dataframe.head()
```

```python
new_dataframe['Tags'][0]
```

#### now we will do text vectorization using bag of words.

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors = cv.fit_transform(new_dataframe['Tags']).toarray()
```

```python
## here most of the words have same origin so we will do stemming here.

import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)
```

```python
new_dataframe['Tags'] = new_dataframe['Tags'].apply(stem)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
## Now we will create a function for recommendation of movies.

def recommend_movies(movie):
    movie_index = new_dataframe[new_dataframe['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)),reverse=True,key = lambda x:x[1])[1:6]

    for i in movies_list:
        print(new_dataframe.iloc[i[0]].title)
```

```python
recommend_movies("Groundhog Day")
```
